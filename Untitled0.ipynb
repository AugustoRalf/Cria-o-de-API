```python
# Importando biblioteca
from sklearn.linear_model import Perceptron

# Dados de entrada
X = [[0, 0], [0, 1], [1, 0], [1, 1]]

# Saídas desejadas
Y = [0, 1, 1, 1]

# Criando e treinando o perceptron
modelo = Perceptron()
modelo.fit(X, Y)

# Testando o modelo
print("Previsões:")
testes = [[0,0], [0,1], [1,0], [1,1]]
for teste in testes:
  previsao = modelo.predict([teste])
  print(f"Nuvens: {teste[0]}, Previsão Chuva: {teste[1]} => Levar Guarda-chuva? {'Sim' if previsao[0] == 1 else 'Não'}")

# Decisão de ir ao parque
from sklearn.linear_model import Perceptron
import numpy as np

# Tabela de Dados: [Tempo úmido, Dia da semana, Parque lotado]
X = np.array([
    [0, 0, 0],  # Não está úmido, dia útil, parque não lotado
    [0, 1, 0],  # Não está úmido, final de semana, parque não lotado
    [1, 0, 0],  # Está úmido, dia útil, parque não lotado
    [1, 1, 0],  # Está úmido, final de semana, parque não lotado
    [0, 0, 1],  # Não está úmido, dia útil, parque lotado
    [0, 1, 1],  # Não está úmido, final de semana, parque lotado
    [1, 0, 1],  # Está úmido, dia útil, parque lotado
    [1, 1, 1]   # Está úmido, final de semana, parque lotado
])

# Saída: 1 = Ir ao Parque, 0 = Não Ir ao Parque
y = np.array([0, 1, 1, 1, 0, 0, 0, 0])

# Criando e treinando o modelo Perceptron
modelo_parque = Perceptron()
modelo_parque.fit(X, y)

# Testando o modelo com novos dados
novo_dia = np.array([[1, 1, 0]])  # Está ensolarado, é final de semana, parque não lotado
previsao = modelo_parque.predict(novo_dia)
print(f"Decisão: {'Ir ao Parque' if previsao[0] == 1 else 'Não Ir ao Parque'}")
# Importando a biblioteca
from sklearn.linear_model import Perceptron

# Dados de entrada
X = [[0, 1, 1, 1], [1, 0, 1, 1], [1, 1, 0, 1], [0, 0, 1, 0], [1, 1, 1, 1], [0, 1, 0, 0], [1, 0, 0, 1], [0, 0, 0, 1]]

# Saídas desejadas
Y = [0, 1, 0, 1, 1, 0, 0, 0]

# Criando e treinando o perceptron
modelo = Perceptron()
modelo.fit(X, Y)

# Testando o modelo
print("Previsões:")
testes = [[0,1,1,1], [1,0,1,1], [1,1,0,1], [0,0,1,0], [1,1,1,1], [0,1,0,0], [1,0,0,1], [0,0,0,1]]
for teste in testes:
    previsao = modelo.predict([teste])
    print(f"Cansado: {teste[0]}, Ingredientes em casa: {teste[1]}, Restaurante aberto: {teste[2]}, Pagamento recente: {teste[3]} => Comer Fora? {'Sim' if previsao[0] == 1 else 'Não'}")
```
